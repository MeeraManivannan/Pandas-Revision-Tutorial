**Matplotlib - The Foundation**

Getting Started

```python
import matplotlib.pyplot as plt
import numpy as np

# Basic configuration

plt.style.use('seaborn-v0_8')  # Modern style
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['font.size'] = 12

print(f"Matplotlib version: {plt.matplotlib.__version__}")
```

Line Plots

```python

# Basic line plot

x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.figure(figsize=(10, 6))

plt.plot(x, y1, label='sin(x)', linewidth=2, color='blue')
plt.plot(x, y2, label='cos(x)', linewidth=2, color='red', linestyle='--')

plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Sine and Cosine Functions')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show
```

```python

# Multiple lines with different styles

x = np.arange(0, 10, 1)

y1 = x
y2 = x**2
y3 = x**3

plt.figure(figsize=(10, 6))

plt.plot(x, y1, 'o-', label='Linear', linewidth=2, markersize=8)
plt.plot(x, y2, 's--', label='Quadratic', linewidth=2, markersize=8)
plt.plot(x, y3, '^:', label='Cubic', linewidth=2, markersize=8)

plt.xlabel('X')
plt.ylabel('Y')
plt.title('Power Functions')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

```

Scatter Plots

```python
# Basic scatter plot

np.random.seed(42)

x = np.random.randn(100)
y = 2 * x + np.random.randn(100) * 0.5

plt.figure(figsize=(10, 6))

plt.scatter(x, y, alpha=0.6, c='blue', s=50)

plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Scatter Plot with Random Data')
plt.grid(True, alpha=0.3)
plt.show()

```

```python

# Colored scatter plot with size variation

n = 200

x = np.random.randn(n)
y = np.random.randn(n)

colors = np.random.randn(n)
sizes = np.random.randint(20, 200, n)

plt.figure(figsize=(10, 6))

scatter = plt.scatter(
    x,
    y,
    c=colors,
    s=sizes,
    alpha=0.6,
    cmap='viridis'
)

plt.colorbar(scatter, label='Color intensity')

plt.xlabel('X')
plt.ylabel('Y')
plt.title('Scatter Plot with Color and Size Variation')

plt.grid(True, alpha=0.3)
plt.show()

```

Bar Charts

```python

# Vertical bar chart

categories = ['A', 'B', 'C', 'D', 'E']
values = [23, 45, 56, 78, 32]

plt.figure(figsize=(10, 6))

bars = plt.bar(
    categories,
    values,
    color='steelblue',
    edgecolor='black'
)

plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Vertical Bar Chart')

# Add value labels on bars

for bar, value in zip(bars, values):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 1,
        str(value),
        ha='center',
        va='bottom'
    )

plt.show()

```

Histograms

```python

# Basic histogram

data = np.random.randn(1000)

plt.figure(figsize=(10, 6))

plt.hist(
    data,
    bins=30,
    color='steelblue',
    edgecolor='black',
    alpha=0.7
)

plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram of Random Data')
plt.grid(True, alpha=0.3, axis='y')
plt.show()

```

```python

# Multiple histograms

data1 = np.random.randn(1000)
data2 = np.random.randn(1000) + 2

plt.figure(figsize=(10, 6))

plt.hist(
    data1,
    bins=30,
    alpha=0.5,
    label='Dataset 1',
    color='blue'
)

plt.hist(
    data2,
    bins=30,
    alpha=0.5,
    label='Dataset 2',
    color='red'
)

plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Overlaid Histograms')
plt.legend()
plt.grid(True, alpha=0.3, axis='y')
plt.show()

```

```python

# Cumulative histogram

plt.figure(figsize=(10, 6))

plt.hist(
    data,
    bins=30,
    cumulative=True,
    density=True,
    color='green',
    edgecolor='black',
    alpha=0.7
)

plt.xlabel('Value')
plt.ylabel('Cumulative Probability')
plt.title('Cumulative Histogram')

plt.grid(True, alpha=0.3)
plt.show()

```

Pie Charts

```python

# Basic pie chart

sizes = [25, 30, 20, 15, 10]

labels = [
    'Category A',
    'Category B',
    'Category C',
    'Category D',
    'Category E'
]

colors = [
    'gold',
    'lightcoral',
    'lightskyblue',
    'lightgreen',
    'pink'
]

explode = (0.05, 0.05, 0.05, 0.05, 0.05)

plt.figure(figsize=(8, 8))

plt.pie(
    sizes,
    explode=explode,
    labels=labels,
    colors=colors,
    autopct='%1.1f%%',
    shadow=True,
    startangle=90
)

plt.title('Distribution by Category')
plt.axis('equal')
plt.show()

```

Donut Chart

```python
# Donut chart

plt.figure(figsize=(8, 8))

wedges, texts, autotexts = plt.pie(
    sizes,
    labels=labels,
    colors=colors,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops={'width': 0.3}
)

plt.title('Donut Chart')
plt.show()
```

Basic Box Plot

```python
# Basic box plot

data = [np.random.randn(100) for _ in range(4)]

plt.figure(figsize=(10, 6))

plt.boxplot(
    data,
    labels=['Group 1', 'Group 2', 'Group 3', 'Group 4']
)

plt.xlabel('Groups')
plt.ylabel('Values')
plt.title('Box Plot Comparison')
plt.grid(True, alpha=0.3, axis='y')
plt.show()
```

Horizontal Box Plot

```python
# Horizontal box plot

plt.figure(figsize=(10, 6))

plt.boxplot(
    data,
    labels=['Group 1', 'Group 2', 'Group 3', 'Group 4'],
    vert=False
)

plt.xlabel('Values')
plt.ylabel('Groups')
plt.title('Horizontal Box Plot')
plt.grid(True, alpha=0.3, axis='x')
plt.show()
```

Multiple Subplots

```python
# Creating multiple subplots

fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Subplot 1: Line plot

x = np.linspace(0, 10, 100)

axes[0, 0].plot(
    x,
    np.sin(x),
    'b-',
    linewidth=2
)

axes[0, 0].set_title('Sine Wave')
axes[0, 0].grid(True, alpha=0.3)

# Subplot 2: Scatter plot

x = np.random.randn(100)
y = np.random.randn(100)

axes[0, 1].scatter(
    x,
    y,
    alpha=0.6,
    c='red'
)

axes[0, 1].set_title('Scatter Plot')
axes[0, 1].grid(True, alpha=0.3)

# Subplot 3: Bar chart

categories = ['A', 'B', 'C', 'D']
values = [23, 45, 56, 78]

axes[1, 0].bar(
    categories,
    values,
    color='green'
)

axes[1, 0].set_title('Bar Chart')
axes[1, 0].grid(True, alpha=0.3, axis='y')

# Subplot 4: Histogram

data = np.random.randn(1000)

axes[1, 1].hist(
    data,
    bins=30,
    color='purple',
    alpha=0.7
)

axes[1, 1].set_title('Histogram')
axes[1, 1].grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()
```

Advanced Matplotlib

```python
# Customizing figure appearance

plt.figure(figsize=(12, 8))

# Data

x = np.linspace(0, 10, 100)
y = np.sin(x)

# Plot with customization

plt.plot(
    x,
    y,
    'b-',
    linewidth=2,
    label='sin(x)'
)

plt.fill_between(
    x,
    y,
    0,
    alpha=0.3,
    color='blue'
)

# Adding annotations

plt.annotate(
    'Maximum',
    xy=(np.pi / 2, 1),
    xytext=(np.pi / 2 + 1, 1.2),
    arrowprops=dict(
        facecolor='black',
        shrink=0.05
    )
)

# Adding reference lines

plt.axhline(
    y=0,
    color='black',
    linestyle='-',
    linewidth=0.5
)

plt.axvline(
    x=np.pi,
    color='red',
    linestyle='--',
    alpha=0.5
)

# Adding text

plt.text(
    7,
    0.5,
    'Signal',
    fontsize=12,
    bbox=dict(
        facecolor='yellow',
        alpha=0.5
    )
)

plt.xlabel('X-axis', fontsize=12)
plt.ylabel('Y-axis', fontsize=12)

plt.title(
    'Advanced Plot with Annotations',
    fontsize=14,
    fontweight='bold'
)

plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

Seaborn Setup

```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Set style

sns.set_style('whitegrid')
sns.set_palette('husl')

# Load built-in datasets

tips = sns.load_dataset('tips')
iris = sns.load_dataset('iris')
titanic = sns.load_dataset('titanic')

print(f"Seaborn version: {sns.__version__}")
print(f"Tips dataset shape: {tips.shape}")
print(f"Iris dataset shape: {iris.shape}")
print(f"Titanic dataset shape: {titanic.shape}")
```

Histogram + KDE

```python
# Distribution plots

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)

sns.histplot(
    tips['total_bill'],
    bins=30,
    kde=True,
    color='blue'
)

plt.title('Total Bill Distribution')

plt.subplot(1, 2, 2)

sns.kdeplot(
    tips['total_bill'],
    fill=True,
    color='green'
)

plt.title('Kernel Density Estimate')

plt.tight_layout()
plt.show()
```

Multiple Distributions

```python
# Multiple distributions

plt.figure(figsize=(10, 6))

for day in tips['day'].unique():

    subset = tips[tips['day'] == day]

    sns.kdeplot(
        subset['total_bill'],
        label=day,
        fill=True,
        alpha=0.3
    )

plt.title('Total Bill Distribution by Day')
plt.xlabel('Total Bill')
plt.ylabel('Density')
plt.legend()
plt.show()
```

Scatter Plot with Regression Line

```python
# Scatter plot with regression line

plt.figure(figsize=(10, 6))

sns.regplot(
    x='total_bill',
    y='tip',
    data=tips,
    scatter_kws={'alpha': 0.5}
)

plt.title('Tip vs Total Bill with Regression Line')
plt.show()
```

Scatter Plot with Hue

```python
# Scatter plot with hue

plt.figure(figsize=(10, 6))

sns.scatterplot(
    x='total_bill',
    y='tip',
    data=tips,
    hue='sex',
    size='size',
    sizes=(20, 200),
    alpha=0.6
)

plt.title('Tip vs Total Bill by Gender and Party Size')

plt.legend(
    bbox_to_anchor=(1.05, 1),
    loc='upper left'
)

plt.tight_layout()
plt.show()
```

Pairplot

```python
# Pairplot for multiple variables

sns.pairplot(
    iris,
    hue='species',
    diag_kind='kde'
)

plt.suptitle(
    'Iris Dataset Pairplot',
    y=1.02
)

plt.show()
```

Seaborn Bar Plot

```python
# Bar plot

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)

sns.barplot(
    x='day',
    y='total_bill',
    data=tips,
    errorbar=95,
    palette='viridis'
)

plt.title('Average Total Bill by Day')

plt.subplot(1, 2, 2)

sns.barplot(
    x='day',
    y='tip',
    data=tips,
    estimator=np.sum,
    palette='plasma'
)

plt.title('Total Tips by Day')

plt.tight_layout()
plt.show()
```

Seaborn Box Plot

```python
# Box plot

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)

sns.boxplot(
    x='day',
    y='total_bill',
    data=tips,
    palette='Set2'
)

plt.title('Total Bill Distribution by Day')

plt.subplot(1, 2, 2)

sns.boxplot(
    x='time',
    y='tip',
    data=tips,
    hue='sex',
    palette='Set3'
)

plt.title('Tip Distribution by Time and Gender')

plt.tight_layout()
plt.show()
```

Violin Plot

```python
# Violin plot

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)

sns.violinplot(
    x='day',
    y='total_bill',
    data=tips,
    palette='muted'
)

plt.title('Violin Plot of Total Bill by Day')

plt.subplot(1, 2, 2)

sns.violinplot(
    x='time',
    y='tip',
    data=tips,
    hue='sex',
    split=True
)

plt.title('Split Violin Plot')

plt.tight_layout()
plt.show()
```

Count Plot

```python
# Count plot

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)

sns.countplot(
    x='day',
    data=tips,
    palette='Set1'
)

plt.title('Number of Meals by Day')

plt.subplot(1, 2, 2)

sns.countplot(
    x='time',
    data=tips,
    hue='sex',
    palette='Set2'
)

plt.title('Meal Times by Gender')

plt.tight_layout()
plt.show()
```

Correlation Heatmap

```python
# Correlation heatmap

numeric_cols = tips.select_dtypes(
    include=[np.number]
)

corr = numeric_cols.corr()

plt.figure(figsize=(10, 8))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    center=0,
    square=True,
    linewidths=1,
    cbar_kws={'shrink': 0.8}
)

plt.title('Correlation Heatmap of Tips Dataset')
plt.show()
```

Clustered Heatmap

```python
# Clustered heatmap

sns.clustermap(
    iris.drop('species', axis=1).corr(),
    annot=True,
    cmap='coolwarm',
    center=0
)

plt.suptitle(
    'Clustered Correlation Heatmap',
    y=1.02
)

plt.show()
```

Facet Grid

```python
# Creating facet grids

g = sns.FacetGrid(
    tips,
    col='time',
    row='sex',
    margin_titles=True
)

g.map(
    sns.scatterplot,
    'total_bill',
    'tip',
    alpha=0.6
)

g.set_axis_labels(
    'Total Bill',
    'Tip'
)

g.set_titles(
    col_template='{col_name}',
    row_template='{row_name}'
)

plt.suptitle(
    'Tips by Time and Gender',
    y=1.02
)

plt.show()
```

Complex Facet Grid

```python
# Complex facet grid

g = sns.FacetGrid(
    tips,
    col='day',
    hue='sex',
    height=5
)

g.map(
    sns.kdeplot,
    'total_bill',
    fill=True,
    alpha=0.5
)

g.add_legend()

g.set_axis_labels(
    'Total Bill',
    'Density'
)

plt.suptitle(
    'Total Bill Distribution by Day and Gender',
    y=1.02
)

plt.show()
```

Joint Plot

```python
# Joint plot

sns.jointplot(
    x='total_bill',
    y='tip',
    data=tips,
    kind='hex',
    marginal_kws=dict(
        bins=30,
        fill=True
    )
)

plt.suptitle(
    'Joint Plot with Hexbin',
    y=1.02
)

plt.show()
```

Pair Grid

```python
# Pair grid

g = sns.PairGrid(
    iris,
    hue='species',
    height=2.5
)

g.map_diag(
    sns.histplot,
    kde=True
)

g.map_offdiag(
    sns.scatterplot,
    alpha=0.5
)

g.add_legend()

plt.suptitle(
    'PairGrid of Iris Dataset',
    y=1.02
)

plt.show()
```

Catplot

```python
# Catplot - flexible categorical plots

g = sns.catplot(
    x='day',
    y='total_bill',
    data=tips,
    kind='box',
    col='time',
    hue='sex',
    height=5,
    aspect=1.2
)

g.set_axis_labels(
    'Day',
    'Total Bill'
)

g.set_titles(
    'Time: {col_name}'
)

plt.suptitle(
    'Box Plots by Day, Time, and Gender',
    y=1.05
)

plt.show()
```

Regression Plots

```python
# Regression plots with confidence intervals

g = sns.lmplot(
    x='total_bill',
    y='tip',
    data=tips,
    hue='sex',
    col='time',
    height=5,
    aspect=1.2,
    scatter_kws={'alpha': 0.5}
)

g.set_axis_labels(
    'Total Bill',
    'Tip'
)

plt.suptitle(
    'Regression Plots by Time and Gender',
    y=1.02
)

plt.show()
```
