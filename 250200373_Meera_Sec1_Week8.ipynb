Q5. What is the average IMDb rating of all movies?

```python
average_rating = df['vote_average'].mean()

print("Average IMDb rating:", average_rating)
```

Q6. What is the median IMDb rating?

```python
median_rating = df['vote_average'].median()

print("Median IMDb rating:", median_rating)
```

Q7. Which five movies have the highest IMDb ratings?

```python
print(df.nlargest(5, 'vote_average')[['title', 'vote_average']])
```

Q8. Which five movies have the highest number of votes?

```python
print(df.nlargest(5, 'vote_count')[['title', 'vote_count']])
```

Q9. Find the average gross revenue for movies grouped by release year.

```python
df['Year'] = pd.to_datetime(df['release_date']).dt.year

average_popularity = df.groupby('Year')['popularity'].mean()

print(average_popularity)
```

Q10. Create a histogram showing the distribution of IMDb ratings.

```python
import matplotlib.pyplot as plt

plt.hist(df['vote_average'], bins=10)

plt.xlabel('IMDb Rating')
plt.ylabel('Number of Movies')
plt.title('Distribution of IMDb Ratings')

plt.show()
```

Q11. Compare the average IMDb rating of movies with more than 100,000 votes against movies with fewer than or equal to 100,000 votes.

```python
more_than_100k = df[df['vote_count'] > 100000]['vote_average'].mean()
less_or_equal_100k = df[df['vote_count'] <= 100000]['vote_average'].mean()

print("Average rating for movies with more than 100,000 votes:",
      round(more_than_100k, 2))

print("Average rating for movies with 100,000 or fewer votes:",
      round(less_or_equal_100k, 2))
```

Q12. Which movie has the highest popularity?

```python
highest_popularity = df.loc[df['popularity'].idxmax(), ['title', 'popularity']]

print(highest_popularity)
```

Q13. Which movie was released most recently?

Q14. Which movie was released earliest?

```python
earliest_movie = df.loc[df['release_date'].idxmin(), ['title', 'release_date']]

print(earliest_movie)
```
